## Week 13 - Text Analytics and Recommendation Techniques

```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

text = "Data Science combines statistics, programming, and domain knowledge to extract insights."
tokens = word_tokenize(text.lower())
stop_words = set(stopwords.words('english'))
filtered = [w for w in tokens if w.isalpha() and w not in stop_words]

stemmer = PorterStemmer()
stemmed = [stemmer.stem(w) for w in filtered]

nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in filtered]

print("Tokens:", filtered)
print("Stemmed:", stemmed)
print("Lemmatized:", lemmatized)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

items = pd.DataFrame({
    'title': ['Action Movie', 'Romantic Comedy', 'Action Thriller', 'Drama Film'],
    'description': ['action fight explosion', 'romance funny love', 'action suspense chase', 'emotional drama story']
})

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(items['description'])
similarity = cosine_similarity(tfidf_matrix)

def recommend(title, top_n=2):
    idx = items[items['title'] == title].index[0]
    scores = list(enumerate(similarity[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)[1:top_n+1]
    return [items['title'][i] for i, _ in scores]

print("Recommendations for 'Action Movie':", recommend('Action Movie'))
```
